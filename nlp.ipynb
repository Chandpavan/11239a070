```python
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
text = ["running", "leaves", "better", "cats"]
lem = WordNetLemmatizer()
for word in text:
    print(word, "→", lem.lemmatize(word))
```

```python
# Install and download necessary data
import nltk
nltk.download('punkt')

# Tokenization
from nltk.tokenize import word_tokenize, sent_tokenize

text = "NLP is fun. It involves text processing."

# Word and sentence tokenization
print("Word Tokenization:", word_tokenize(text))
print("Sentence Tokenization:", sent_tokenize(text))
```
